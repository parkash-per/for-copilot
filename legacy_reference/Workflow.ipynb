# Satellite Altimetry - Calibration/Validation Moorings in Bass Strait

Data acquired from the moored instruments is downloaded from the instruments by NCMI technicians and provided to CSIRO staff via SSD and mutually accessible server. Pre-processing Quality Assurance checks are conducted to ensure all data has been successfully downloaded from instruments, metadata is correct, the correct file name(s) have been assigned to files, and data has been archived in multiple locations by NCMI and CSIRO.

Main code:
<ul style="padding-left: 80px;">
    <li>proc_aqd.ipynb, qc_aqd.ipynb, imos_aqd.ipynb</li>
    <li>proc_mat1.ipynb, qc_mat1.ipynb, imos_mat1.ipynb</li>    
    <li>proc_rbrq.ipynb, qc_rbrq.ipynb, imos_rbrq.ipynb</li>
    <li>proc_sbe26.ipynb, qc_sbe26.ipynb, imos_sbe26.ipynb</li>
    <li>proc_sbe37.ipynb, qc_sbe37.ipynb, imos_sbe37.ipynb</li>
    <li>proc_sig500.ipynb, qc_sig500.ipynb, imos_sig500.ipynb</li>
</ul>

Tools:
<ul style="padding-left: 80px;">
    <li>parsers - for each instrument there is a parser used in the proc code</li>
    <li>helpers - generic, frequently used bits of code</li>
    <li>imos_nc_converter - converts each instrument to IMOS-compliant netCDF, defines global attributes and contains a class for each instrument type to define variables and their attributes</li>
    <li>tide_current_analysis_predict - tools for tide and current analysis and prediction</li>
</ul>

## Data Processing

Processing of instrument data is ultimately to provide IMOS compliant files for end users, utilising a combination of third-party software and CSIRO-developed code.


1. Ensure correct metadata in database
/datasets/work/oa-srsalt/work/preqa/SWOT/cal_val/jason_calval/all_mooring_data/reference_mooring_proc_info/satellite_altimetry_moorings_metadata.csv
<ul style="padding-left: 80px;">
    <li>Time offset information, if provided by MIT with data</li>
    <li>Paths, i.e. data_in_path, proc_1_path, etc</li>
    <li>Nominal depth, mooring and instrument</li>
</ul>

2. Process Mooring's pressure sensor - RBRQ or SBE26. 
<ul style="padding-left: 80px;">
    <li>proc_rbrq.ipynb</li>
    <li>proc_sbe26.ipynb</li>
    <li><i>CHECK IF PRESSURE CORRECTION OFFSET NEEDED, APPLY IF NECESSARY</i>
</ul>

----------------------------------------------------------------------------------------------------------------------
<pre>
        <i> Use the start/end of the valid pressure data to determine 'time_coverage_start' and 'time_coverage_end' for <b>ALL</b> instruments on the mooring.
